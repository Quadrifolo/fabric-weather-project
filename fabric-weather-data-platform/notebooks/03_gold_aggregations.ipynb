```python
# Welcome to your new notebook
# Type here in the cell editor to add code!
from pyspark.sql import functions as F

silver_df = spark.table("silver_weather_observations")
```

```python
gold_daily = (
    silver_df
    .withColumn(
        "observation_date",
        F.to_date("observation_time")
    )
    .groupBy("observation_date")
    .agg(
        F.round(F.avg("temperature_2m"), 2).alias("avg_temperature"),
        F.round(F.min("temperature_2m"), 2).alias("min_temperature"),
        F.round(F.max("temperature_2m"), 2).alias("max_temperature"),
        F.round(F.avg("relative_humidity_2m"), 2).alias("avg_humidity"),
        F.round(F.sum("precipitation"), 2).alias("total_precipitation"),
        F.round(F.avg("wind_speed_10m"), 2).alias("avg_wind_speed"),
        F.round(F.max("wind_speed_10m"), 2).alias("max_wind_speed")
    )
    .orderBy("observation_date")
)
```

```python
display(gold_daily)
```

```python
gold_daily.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("gold_daily_weather")
```
