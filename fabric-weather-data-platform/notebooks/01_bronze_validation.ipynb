```python
raw_df = (
    spark.read
    .option("multiline", "true")
    .json("Files/bronze/weather/raw/*.json")
)

raw_df.printSchema()
display(raw_df)
```

```python
required_columns = ["hourly"]

missing_columns = [
    col for col in required_columns
    if col not in raw_df.columns
]

if missing_columns:
    raise Exception(
        f"Bronze validation failed. Missing columns: {missing_columns}"
    )

print("Bronze top-level validation passed.")
```

```python
required_hourly_fields = [
    "time",
    "temperature_2m",
    "relative_humidity_2m",
    "precipitation",
    "wind_speed_10m"
]

hourly_fields = raw_df.schema["hourly"].dataType.fieldNames()

missing_hourly = [
    field for field in required_hourly_fields
    if field not in hourly_fields
]

if missing_hourly:
    raise Exception(
        f"Bronze validation failed. Missing hourly fields: {missing_hourly}"
    )

print("Bronze hourly structure validation passed.")
```
