```python
from delta.tables import DeltaTable
from datetime import timedelta

watermark_table = DeltaTable.forName(
    spark,
    "pipeline_watermark"
)

current_watermark = (
    spark.table("pipeline_watermark")
    .filter("pipeline_name = 'weather_ingestion'")
    .select("last_successful_date")
    .first()[0]


new_watermark = current_watermark + timedelta(days=1)

watermark_table.update(
    condition="pipeline_name = 'weather_ingestion'",
    set={
        "last_successful_date": f"DATE('{new_watermark}')"
    }
)

print(f"Watermark updated: {current_watermark} -> {new_watermark}")
```
